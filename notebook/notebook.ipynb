```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 24, 34, 44, 54, 64, 100],
    labels=["Under 25", "25-34", "35-44", "45-54", "55-64", "65+"]
)
```

```python
df["Application_Year"] = df["Application_Date"].dt.year
```

```python
df["Application_Year"].value_counts().sort_index()
```

```python
df["Loan_Security"] = df["Collateral_Type"].apply(
    lambda x: "Unsecured" if pd.isna(x) else "Secured"
)
```

```python
df.columns
```

```python
df.head(10)
```

```python
df["Credit_Score_Band"] = pd.cut(
    df["Credit_Score"],
    bins=[0, 579, 669, 739, 799, 900],
    labels=["Poor", "Fair", "Good", "Very Good", "Excellent"],
    include_lowest=True
)
```

```python
df["DTI_Band"] = pd.cut(
    df["DTI_Ratio_Pct"],
    bins=[0, 20, 35, 50, 65, 100],
    labels=["Low", "Moderate", "High", "Very High", "Extreme"],
    include_lowest=True
)
```

```python
df["LTV_Band"] = pd.cut(
    df["LTV_Pct"],
    bins=[0, 60, 80, 90, 100],
    labels=["Low", "Moderate", "High", "Very High"],
    include_lowest=True
)
```
